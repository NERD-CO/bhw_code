```python
#####
# This is a test script recreating "NWB_NPIX_BASICPROC.m" matlab script from Dan Kramer. 
# Prior to this, create a venv with python and the pynwb package, then add the NPIX raw data file.
# Remember when translating matlab to python, matlab counts from 1 but py counts from 0.
import pynwb

# This points to the .nwb data file
nwb_file_path = "/Users/bhwong/Desktop/PROJECTS/npix-data/NPIX_MSIT_004.nwb"

# This defines the reader, which is the NWBHDF5IO class, and the mode "read"
io = pynwb.NWBHDF5IO(nwb_file_path, mode='r')

# nwb variable instructs to read data file
nwbObj = io.read()

print(nwbObj)
```

```python
# This prints the trials table, which is a pandas dataframe that contains the trial information for each trial in the experiment. 
# colnames are the headers, columns in this case are data containers, each one in vectordata format
# vectordata format is how pynwb stores data
print(nwbObj.trials)

# Copied from DRK NPIX_BASICPROC file:
# Condition = a conflict trial, 1 = a non conflict trial (xx3 or 100)
# Conflict = ???
# Correct = what the correct answer is supposed to be
# ISI = fixation cross time of the preceding ITI 
# ResponseAccuracy = if the correct answer is the same as what they pressed (so correct or not)
# ResponseKey = what they actually pressed
# ResponseUncertainty = ???
# Stimulation = ???
# Stimuli = the stimuli the patient saw
# Trial = trial number, 64 trial blocks
# fixation_time = start of the iti in ms
# responsetime = when they responded
# stimulus_time = when the stimulus appeared
# trial_end_time = when the trial ends and is fractions of a ms from when the next fixation time comes on.
```

```python
# Creates a pandas dataframe from the trials table, using .to_dataframe()
# keep in mind that trials only go something like 1-64 since the MSIT is done in short runs (?)
# the ID represents the total number of lines of data.
trials_df = nwbObj.trials.to_dataframe()

# Optional, set pandas to not truncate the table
import pandas as pd
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)

print(trials_df.to_string())

# depth, xpos, ypos, shank, channel are the physical location of each recording site on probe
# label, KSlabel are quality labels, KS is KiloSort
# KSamplitude KScontamination are KS quality metrics
# n_spikes and amp are number of spikes and average amplitude 
# waveform mean is average spike shape
# spike_times is timestamp of each spike, in one long concatenated list
# spike_times_index is a list of cutoff points per neuron/recording, 
# e.g., [356, 891, etc] as an index means neuron 0 contains spike times 0-356, neuron 1 contains 357-891, etc.
```

```python
# Now print units table
# depth, xpos, ypos, shank, channel are the physical location of each recording site on probe
# label, KSlabel are quality labels, KS is KiloSort
# KSamplitude KScontamination are KS quality metrics
# n_spikes and amp are number of spikes and average amplitude 
# waveform mean is average spike shape
# spike_times is timestamp of each spike, in one long concatenated list
# spike_times_index is a list of cutoff points per neuron/recording, 
# e.g., [356, 891, etc] as an index means neuron 0 contains spike times 0-356, neuron 1 contains 357-891, etc.

print(nwbObj.units)

```

```python
# Create pandas dataframe from units table
# Keep in mind that each row is a different neuron/recording, while spike_times for each recording represents its own list of spike times.
# The spike times that correlate to each recording row is automatically unpacked by the to_dataframe() function, based on the spike_times_index!

units_df = nwbObj.units.to_dataframe()
print(units_df.to_string())
```

```python
# If you want to access the spike times for a specific neuron/recording, you can use this function:
# units_df.loc[0, 'spike_times']

# Next, we will sort by label (good, mua, noise)
# We call the units_df that we just created (this contains the dataframe)
# Convention is dataframe['column'].method()

units_df['label'].value_counts()


```

```python
# Now we can filter to remove the noise, keeping only good and MUA
# Convention to sort into Boolean array is: dataframe['column'] where != means "not equal to" 
# Because just calling units_df['label'] != 'noise' will give us a boolean array of True and False,
# we have to call units_df[units_df['label'] != 'noise']. 
# This is saying "call units_df, but only parts of the df where, label != noise"

units_clean_df = units_df[units_df['label'] != 'noise']

print(units_clean_df)

# the len() function tells us number of rows
# the f string lets you print text and variables together
# anything within {} in the f string is evaluated as code

print(f"Total rows: {len(units_df)}")
```
